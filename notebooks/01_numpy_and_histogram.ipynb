```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([1, 2, 3, 4, 5])
print(a * 2, np.mean(a), np.std(a))
```

```python
signal = np.random.normal(3.1, 0.05, 10_000)
background = np.random.uniform(2.5, 3.7, 20_000)
data = np.concatenate([signal, background])

plt.hist(data, bins=100)
plt.xlabel("Mass [GeV]")
plt.ylabel("Events")
plt.title("Toy data: signal peak on flat background")
plt.show()
```
